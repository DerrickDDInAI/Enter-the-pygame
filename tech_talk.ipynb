```python
def collide(gorilla, gorilla_x, gorilla_y, obstacle, obstacle_x, obstacle_y):
    """
    returns true if collision
    :param: gorilla
    :return: Bool
    """
    gorilla_mask = pygame.mask.from_surface(gorilla)
    obstacle_mask = pygame.mask.from_surface(obstacle)
    offset = (obstacle_x - gorilla_x, obstacle_y - gorilla_y)

    collision = gorilla_mask.overlap(obstacle_mask, offset)

    if collision:
        return True

    else:
        return False

def draw_text(text, color, xy_pos_center_px, screen) -> None:
    """
    Function to display text.

    Parameters:
    * text: text to display
    * color: font color
    * xy_pos_center_px: (x,y) coordinates of the center of text rectangle.
    """
    text_surface = game_font.render(
        text, True, color)  # text, True for anti-aliased text, color in rgb
    
    # Put the text in a rect to make it easier to display
    text_rect = text_surface.get_rect(center=xy_pos_center_px)

    # Draw the text on the game_window screen
    screen.blit(text_surface, text_rect)
```

```python
# CONSTANTS
WINDOW_WIDTH_PX = 1440
WINDOW_HEIGHT_PX = 900

# Variables
caption = "Tech Talk"

bg_surface_imagepath = "gamecore/assets/images/mars_sand_dunes.jpeg"
gorilla_imagepath = "gamecore/assets/images/gorilla.png"
gorilla_ss_imagepath = "gamecore/assets/images/gorilla_ss.png"
vegegorilla_imagepath = "gamecore/assets/images/vegegorilla.png"
asteroid_imagepath = "gamecore/assets/images/asteroid.png"

gorilla_soundpath = "gamecore/assets/sounds/gorilla_sounds.mp3"
genkidama_soundpath = "gamecore/assets/sounds/genkidama.mp3"
sfx_collision_soundpath = "gamecore/assets/sounds/sfx_hit.wav"

running = True
game_active = True
time_s = 0.0
framerate_limit = 120
ss_bar = 0.0
becode_color = (22, 35, 46)
transition_color = [0,0,0]
alpha = 255
alpha_ss = 0
alpha_vg = 0
text = ""


gorilla_x = 50
gorilla_y = 50
gorilla_dir = "left"
asteroid_x = 600
asteroid_y = 0
play_ss_sound = False

# Setup
pygame.init() # Initialize pygame
pygame.font.init() # Initialize font
game_font = pygame.font.SysFont("comicsans", 50) # Setting font and size

# Initiate recognition and talking
listener = sr.Recognizer()
engine = pyttsx3.init()

# Display
## Create a display surface called screen by convention
screen = pygame.display.set_mode((WINDOW_WIDTH_PX,WINDOW_HEIGHT_PX))
pygame.display.set_caption(caption)



# Clock
main_clock = pygame.time.Clock()  # instantiate clock to limit the frame rate

# Import images
## Import the background image and assign it to a variable
## convert() is not necessary but converts the image into a format easier for pygame => faster
bg_surface = pygame.image.load(bg_surface_imagepath).convert()
bg_surface = pygame.transform.scale(bg_surface, (WINDOW_WIDTH_PX, WINDOW_WIDTH_PX))
asteroid = pygame.image.load(asteroid_imagepath).convert_alpha()
asteroid = pygame.transform.scale(asteroid, (200,200))

gorilla = pygame.image.load(gorilla_imagepath).convert_alpha()
# gorilla = pygame.transform.scale(gorilla, (300,300))
gorilla_flipped = pygame.transform.flip(gorilla, True, False) # flip the image horizontally

gorilla_ss = pygame.image.load(gorilla_ss_imagepath).convert_alpha()
# gorilla_ss = pygame.transform.scale(gorilla_ss, (300,300))
gorilla_ss_flipped = pygame.transform.flip(gorilla_ss, True, False) # flip the image horizontally


vegegorilla = pygame.image.load(vegegorilla_imagepath).convert_alpha()
# vegegorilla = pygame.transform.scale(vegegorilla, (300,300))
vegegorilla_flipped = pygame.transform.flip(vegegorilla, True, False) # flip the image horizontally

# Import sounds
gorilla_sound = pygame.mixer.Sound(gorilla_soundpath)
genkidama = pygame.mixer.Sound(genkidama_soundpath)
collision_sound = pygame.mixer.Sound(sfx_collision_soundpath)

# Game loop
while running:
    # Get the delta t for one frame (this changes depending on system load).
    dt_s: float = float(main_clock.tick(framerate_limit) * 1e-3)
    
    ## Get user input

    with sr.Microphone() as source:
        print ("listening\n")
        audio = listener.listen(source)
        try:
            command = listener.recognize_google(audio)
            print(command)
        except:
            print('Did not catch that')
            command=''

        if 'left' in command:
            gorilla_x -= 300
        if 'right' in command:
            gorilla_x += 300
        if 'up' in command:
            gorilla_y -= 300
        if 'down' in command:
            gorilla_y += 300
        if 'say' in command:
            text = command.replace("say", "")
                    
    # For event in the event queue since last call
    for event in pygame.event.get():

        # Check keys released for one instant
        if event.type == pygame.KEYUP:
            # Quit the game if you release the escape button
            if event.key == pygame.K_ESCAPE:
                running = False
                pygame.quit()
                sys.exit()
            if event.key == pygame.K_RETURN and game_active == False:
                game_active = True
                gorilla_x = 0
                gorila_y = 0
        
        # Check keys pressed for one instant
        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_LEFT:
                gorilla_x -= 45
                gorilla_dir = "left"
            if event.key == pygame.K_RIGHT:
                gorilla_x += 45
                gorilla_dir = "right"
            if event.key == pygame.K_UP:
                gorilla_y -= 45
            if event.key == pygame.K_DOWN:
                gorilla_y += 45

        # Quit the game if you click on the X button at the top of the screen
        if event.type == pygame.QUIT:
            running = False
            pygame.quit()
            sys.exit()
    
    if game_active:
        
        # Check keys continuously pressed (needs to be outside of the for loop otherwise it would only be executed once per event in the event queue)
        keys = pygame.key.get_pressed()
        if keys:
            if keys[pygame.K_SPACE]:
                ss_bar += 0.01

        if collide(gorilla, gorilla_x, gorilla_y, asteroid, asteroid_x, asteroid_y):
            collision_sound.play()
            game_active = False

        ## Draw

        # if super saiyan transformation bar >= 5
        if ss_bar >= 2:
            # Turn the screen whiter every second
            if time_s > 0.01:
                for i, color in enumerate(transition_color):
                    if color < 255:
                        transition_color[i] += 1
                        alpha -= 1
                        # alpha_ss += 1
                time_s = 0.0 # reset timer
            screen.fill(tuple(transition_color))

            if alpha > 0:
                gorilla.set_alpha(alpha)
                gorilla_flipped.set_alpha(alpha)

                if gorilla_dir == "left":
                    screen.blit(gorilla, (gorilla_x, gorilla_y))

                elif gorilla_dir == "right":
                    screen.blit(gorilla_flipped, (gorilla_x, gorilla_y))

            elif alpha < 0 and alpha_ss < 255:
                alpha_ss += 1
                gorilla_ss.set_alpha(alpha_ss)
                gorilla_ss_flipped.set_alpha(alpha_ss)

            else: 
                if gorilla_dir == "left":
                    screen.blit(gorilla_ss, (gorilla_x, gorilla_y))

                elif gorilla_dir == "right":
                    screen.blit(gorilla_ss_flipped, (gorilla_x, gorilla_y))

        else:
            ### To display a surface on another surface: here to display bg_surface on screen
            ### (0,0): the position of the top left of bg_surface
            ### why? (0,0) is the origin
            screen.blit(bg_surface, (0,0))

            if gorilla_dir == "left":
                screen.blit(gorilla, (gorilla_x, gorilla_y))

            elif gorilla_dir == "right":
                screen.blit(gorilla_flipped, (gorilla_x, gorilla_y))

        screen.blit(asteroid, (asteroid_x, asteroid_y))
        
        if len(text) > 1:
            draw_text(text, (128,128,128), (gorilla_x, gorilla_y - 200), screen)
            
        ## Sounds
        if ss_bar > 1 and not play_ss_sound:
            play_ss_sound = True
            genkidama.play()
    
    else:
        screen.fill((0,0,0))
        draw_text("game over", (255,255,255), (WINDOW_WIDTH_PX/2, WINDOW_HEIGHT_PX/2), screen)
        if time_s > 0.01 and alpha_vg < 255:
            alpha_vg += 1
            time_s = 0.0 # reset timer

        vegegorilla.set_alpha(alpha_vg)
        screen.blit(vegegorilla, (500, 200))
    
    pygame.display.update() # Update the screen with the drawings
    ## Time
    ## Measure time spent at each iteration
    time_s += dt_s
    main_clock.tick(framerate_limit) # Limit the frame rate to max the framerate_limit
```
